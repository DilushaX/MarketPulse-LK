#### 1. Import libraries

```python
import os
import pandas as pd
import numpy as np

# Display settings
pd.set_option('display.max_columns', 30)
pd.set_option('display.width', 1000)
```

#### 2. Load raw datasets
Loading raw CSV files for HAYL, JKH, and VONE from data/raw/ without modifying them.


```python
# File paths
files = {
    'HAYL': os.path.join('..', 'data', 'raw', 'HAYL_2014_2026.csv'),
    'JKH':  os.path.join('..', 'data', 'raw', 'JKH_2014_2026.csv'),
    'VONE': os.path.join('..', 'data', 'raw', 'VONE_2014_2026.csv')
}

dfs = {name: pd.read_csv(path) for name, path in files.items()}

for name, df in dfs.items():
    print(f'{name} shape: {df.shape}')
```

#### 3. Inspect columns and data types
Checking whether all 3 datasets share the same schema.


```python
# Compare columns
cols_match = (dfs['HAYL'].columns.tolist() == dfs['JKH'].columns.tolist() == dfs['VONE'].columns.tolist())
print(f'Columns match across all 3 files: {cols_match}')

# Show data types
types_df = pd.DataFrame({
    'Column': dfs['JKH'].columns,
    'HAYL_type': dfs['HAYL'].dtypes.values,
    'JKH_type': dfs['JKH'].dtypes.values,
    'VONE_type': dfs['VONE'].dtypes.values
})
types_df
```

#### 4. Date range and sorting
Check start/end dates and make sure dates are sorted in ascending order.


```python
for name, df in dfs.items():
    dates = pd.to_datetime(df['Date'])
    print(f'{name}: Start = {df["Date"].min()} | End = {df["Date"].max()} | Sorted = {dates.is_monotonic_increasing}')
```

#### 6. Dataset Structure Summary (Table 1)
Save summary to outputs/tables/table1_dataset_structure.csv.


```python
summary = []
for name, df in dfs.items():
    mem_kb = df.memory_usage(deep=True).sum() / 1024
    summary.append({
        'Company': name,
        'Rows': len(df),
        'Columns': len(df.columns),
        'Start Date': df['Date'].min(),
        'End Date': df['Date'].max(),
        'Memory (KB)': round(mem_kb, 2)
    })

table1 = pd.DataFrame(summary)
os.makedirs(os.path.join('..', 'outputs', 'tables'), exist_ok=True)
table1.to_csv(os.path.join('..', 'outputs', 'tables', 'table1_dataset_structure.csv'), index=False)
table1
```

- Date range is 2014-01-02 to 2026-10-14, properly sorted.
- The  column appears completely empty (NaN) across all datasets.
